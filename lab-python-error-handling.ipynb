def initialize_inventory(products):

````python
#numero1
products = ["t-shirt", "mug", "hat", "book", "keychain"]
def initialize_inventory(products):
    inventory = {}
    for product in products:
        valid_input = False
        while not valid_input:
            try:
                quantity = int(input(f"Enter the quantity of {product}s available: "))
                if quantity > 0:
                    inventory[product] = quantity
                    valid_input = True
                else:
                    print("Quantity cannot be negative. Please enter a valid quantity.")
            except ValueError:
                print("Invalid input. Please enter a valid quantity.")
    return inventory
````

````python
#numero 2 
products = ["t-shirt", "mug", "hat", "book", "keychain"]

def calculate_total_price(products):
    precio_inventory = {}
    for i in products:
        valid_input = False
        while not valid_input:
            try:
                precio = float(input(f"Ingrese el precio de {i}: "))
                if precio > 0:
                    precio_inventory[i] = precio
                    valid_input = True
                elif precio == 0:
                    print(f"El precio de {i} no puede ser 0.")
                else:
                    print("El precio no puede ser negativo. Ingrese un precio válido.")
            except ValueError:
                print("Entrada inválida. Ingrese un precio válido.")
            except Exception:
                print("Error inesperado!")
    return precio_inventory
````

````python
# numero 3
def get_customer_orders(inventory):
    orders = []
    valid_input = False
    while not valid_input:
        try:
            rango = int(input("Ingrese cuantos pedidos va a realizar: "))
            if rango > 0:
                valid_input = True
        except ValueError:
            print("Cantidad numérica por favor.")

    pedidos_agregados = 0
    while pedidos_agregados < rango:
        producto = input(f"Ingrese el nombre del producto #{pedidos_agregados + 1}: ")

        try:
            if inventory[producto] <= 0:
                print(f"No hay stock disponible de '{producto}'.")
            else:
                orders.append(producto)
                inventory[producto] -= 1
                pedidos_agregados += 1
        except KeyError:
            print(f"'{producto}' no es un producto válido. Escriba uno de la lista: {list(inventory.keys())}")
        except Exception:
            print("Ocurrió un error inesperado.")

    return orders
````
